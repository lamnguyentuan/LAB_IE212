# **Câu 2: Thống kê tổng số đơn hàng, số lượng khách hàng và người bán.**

```python
#Tổng số đơn hàng
totalOrders = orders_Df.count()
print("Tổng số đơn hàng: ", totalOrders)

#So lượng khách hàng
totalCustomers = customerList_Df.select("Customer_Trx_ID").distinct().count()
print("Số khách hàng: ", totalCustomers)

#Số lượng người bán
totalSellers = orderItems_Df.select("Seller_ID").distinct().count()
print("Số người bán: ", totalSellers)
```

# **Câu 3: Phân tích số lượng đơn hàng theo quốc gia, sắp xếp theo thứ tự giảm dần.**

```python
ordersByCountry = orders_Df.join(customerList_Df, on='Customer_Trx_ID', how='inner')
ordersByCountry.show(5)
```

```python

orders_per_country = ordersByCountry.groupBy("Customer_Country")\
                                    .count()\
                                    .orderBy("count", ascending = False)

orders_per_country.show(5)
```

# **Câu 4:Phân tích số lượng đơn hàng nhóm theo năm, tháng đặt hàng (Hiển thị theo năm tăng dần, tháng giảm dần)**

```python
orders_Df_casted = orders_Df.withColumn("Order_Purchase_Timestamp_Casted", to_timestamp(col("Order_Purchase_Timestamp"), "M/d/yyyy H:mm"))
orders_Df_casted.printSchema()
```

```python
orders_Df_casted = orders_Df_casted.drop("Order_Purchase_Timestamp")
orders_Df_casted.show(5)
```

```python

orders_with_ym = orders_Df_casted.withColumn('year', year(col('Order_Purchase_Timestamp_Casted')))\
                                                   .withColumn('month', month(col('Order_Purchase_Timestamp_Casted')))
orders_with_ym = orders_with_ym.filter(col('year').isNotNull())
orders_grouped = orders_with_ym.groupBy('year', 'month').count().orderBy(col('year').asc(), col('month').desc())
orders_grouped.show(20, truncate=False)
```

# **Câu 5: Thống kê điểm đánh giá trung bình, số lượng đánh giá theo từng mức (ví dụ: 1 đến 5).**

```python
spark.conf.set("spark.sql.ansi.enabled", "false")


reviews_Df = (
    orderReviews_Df
    .withColumn("Review_Score_int", col("Review_Score").cast("int"))
    .filter(col("Review_Score_int").between(1, 5))
    .groupBy("Review_Score_int")
    .count()
    .orderBy("Review_Score_int")
)

reviews_Df.show()
```

```python
avg_review_score = orderReviews_Df.agg(avg(col("Review_Score")\
                                           .cast("float"))\
                                            .alias("Average_Review_Score"))
avg_review_score.show()
```

# **Câu 6: Tính doanh thu (giá sản phẩm + phí vận chuyển) trong năm 2024 và nhóm theo danh mục sản phẩm**


```python
orders_2024 = orders_with_ym.filter(col('year') == 2024)
orders_2024.show(5)
```

```python
orders_items_2024 = orders_2024.join(orderItems_Df, orders_2024["Order_ID"] == orderItems_Df["Order_ID"], how='inner')
orders_items_2024.show(5)
```

```python
orders_items_products_2024 = orders_items_2024.join(products_Df, orders_items_2024["Product_ID"] == products_Df["Product_ID"], how='inner')
orders_items_products_2024.show(5)
```

```python
from pyspark.sql import functions as F

revenue_by_category = orders_items_products_2024.withColumn('Total_Price', F.col('Price')+F.col('Freight_Value')).groupBy('Product_Category_Name')\
                                                .agg(F.sum('Total_Price').alias('Total_Revenue'))
revenue_by_category.show(10)
```
